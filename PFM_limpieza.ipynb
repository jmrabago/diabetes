```python
# Librerías a importar
import pandas as pd
import numpy as np
```

```python
# Lectura del dataset en formato CSV 
df = pd.read_csv('../Trabajo Fin de Master/2023 BRFSS Survey Data.csv')
# Información general del dataset original
df.info()
```

```python
# Se crea otro dataset especialmente con las variables a utilizar
diabetes_dataset = df[['GENHLTH', 'SEXVAR', '_RACE', '_AGE_G', '_INCOMG1', 'EMPLOY1', '_EDUCAG', '_RFHYPE6', '_RFCHOL3',
         'DIFFWALK', 'BLIND', '_SMOKER3', '_RFDRHV8', '_BMI5', 'MENTHLTH', 'PHYSHLTH', 'CVDCRHD4','EXERANY2', 
          '_HLTHPL1', 'DECIDE', 'CHCKDNY2', 'DIABETE4']]

# Se crea un archivo CSV con las variables a analizar
diabetes_dataset.to_csv('diabetes_dataset.csv', index=False)
```

```python
# Información general de la tabla creada con las variables a analizar
diabetes_dataset.info()
```

```python
# Se ve la forma de la tabla (filas, columnas)
diabetes_dataset.shape
```

```python
# Sacamos la suma de las variables nulas
diabetes_dataset.isnull().sum()
```

```python
# Método para obtener el conteo de los distintos valores de cada variable
diabetes_dataset['_INCOMG1'].value_counts()
```

```python
# Eliminar las filas con respuestas indeseadas
diabetes_dataset.drop(diabetes_dataset[diabetes_dataset['_RFDRHV8'] == 9.0].index, axis=0, inplace=True)
```

```python
diabetes_dataset.shape
```

```python
# Se guarda un archivo CSV con el dataset limpio
diabetes_dataset.to_csv('diabetes_dataset_clean.csv', index=False)
```

```python
# Para llenar las variables nulas de una columna del dataset

# Creamos un nuevo dataset por columna
columna = df['_RACE']
# Sacamos la moda de la columna que es el valor que más se repite
mode = columna.mode()
mode
```

```python
# Rellenamos los valores nulos con la moda
diabetes_dataset['_RACE'].fillna(mode[0], inplace=True)
```

```python
diabetes_dataset.dropna(inplace=True)
```

```python
# Quitar los 2 decimales de cada valor de la columna
diabetes_dataset['_BMI5'] = diabetes_dataset['_BMI5'] / 100
```
