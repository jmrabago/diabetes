```python
# Librerías a importar
import pandas as pd
import numpy as np
from sklearn import linear_model, svm, tree, cluster
from sklearn.model_selection import train_test_split
from sklearn import ensemble
from sklearn.metrics import mean_squared_error, r2_score
from sklearn.preprocessing import StandardScaler
from sklearn.neighbors import KNeighborsRegressor
```

```python
# Lectura del dataset
df = pd.read_csv('../Trabajo fin de Master/diabetes.csv')
```

```python
# Forma del dataset (filas, columnas)
df.shape
```

```python
# Nombre de las columnas
df.columns
```

```python
# Array con los nombres de las columnas menos diabetes
features = ['salud_gral', 'sexo', 'raza_etnicidad', 'edad', 'ingreso_anual',
       'empleado', 'nvl_educacion', 'presion_alta', 'colesterol_alto',
       'dif_caminar', 'ciego_o_dif_ver', 'fumador', 'alcoholismo',
       'indice_masa_corporal', 'salud_mental', 'salud_fisica', 'enf_coronaria',
       'act_fisica', 'seguro_gastos_medicos', 'dif_concentrarse',
       'enf_riñones']

# Variable con la variable objetivo
target = 'diabetes'
```

```python
# Inicializamos el objeto scaler
scaler = StandardScaler()

# Elegimos el modelo a utilizar
models = [linear_model.LinearRegression(), linear_model.Ridge(alpha=0.5), linear_model.Lasso(alpha=0.1), 
         linear_model.ElasticNet(random_state=0), linear_model.LogisticRegression(random_state=0, solver='lbfgs', multi_class='multinomial'), 
          linear_model.SGDRegressor(), svm.LinearSVR(), tree.DecisionTreeRegressor(), ensemble.RandomForestRegressor(), 
          ensemble.AdaBoostRegressor(), ensemble.GradientBoostingRegressor(), KNeighborsRegressor()]
```

```python
# Separamos el dataframe entre las variables con las que vamos a entrenar y la variable objetivo
x = scaler.fit_transform(df.loc[:,features].values)
#x = df.loc[:,features].values
#x = df[features]
y = df[target]
```

```python
# Separamos el dataframe en 2: con el que vamos a entrenar y con el que vamos a realizar una prueba
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)
```

```python
# Bucle para usar todos los modelos seleccionados y sacar los errores de cada uno
for model in models:
    print(f'Modelo: {model}') # Se imprime el modelo que se usa
    # Se realiza el entrenamiento del modelo
    model = model.fit(x_train, y_train)
    # Se realiza la prediccion con el dataframe de prueba
    y_pred = model.predict(x_test)
    # Chequeo por r2_score
    variance = r2_score(y_test, y_pred) # Se imprime el valorde la varianza del modelo
    print(f'Varianza del modelo: {variance:.2f}')
    # Chequeo por la media del error cuadratic
    rmse = mean_squared_error(y_pred=y_pred, y_true=y_test, squared=False)
    print(f'RMSE (sesgo) del modelo: {rmse:.2f}\n') # Se imprime el valor del sesgo del modelo
```
